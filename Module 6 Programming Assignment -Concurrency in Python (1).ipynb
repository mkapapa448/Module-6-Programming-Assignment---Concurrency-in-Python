```python
from datetime import date
import time
fmt = "%Y-%m-%d"
now = date.today()
with open("today.txt", "w") as file:
    file.write(now.isoformat())
```

```python
with open("today.txt", "r") as file:
    today_string = file.read()
```

```python
str_time = time.strptime(today_string, fmt)
date_time_now = date(str_time.tm_year, str_time.tm_mon, str_time.tm_mday)
```

```python
import multiprocessing as mp
from datetime import datetime
import random
import time

def random_time():
    wait_time = random.random()
    time.sleep(wait_time)
    print(datetime.now())

if __name__ == '__main__':
    processes = []
    for i in range(3):
        processes.append(mp.Process(target=random_time))
    for i in processes:
        i.start()
    for i in processes:
        i.join()
```
