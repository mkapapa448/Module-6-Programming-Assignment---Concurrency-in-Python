```python
from datetime import date
import time
fmt = "%Y-%m-%d"
now = date.today()
with open("today.txt", "w") as file:
    file.write(now.isoformat())
```

```python
with open("today.txt", "r") as file:
    today_string = file.read()
```

```python
str_time = time.strptime(today_string, fmt)
date_time_now = date(str_time.tm_year, str_time.tm_mon, str_time.tm_mday)
```

```python
import multiprocessing as mp
from datetime import datetime
import random
import time

def random_time():
    wait_time = random.random()
    time.sleep(wait_time)
    print(datetime.now())

time_queue = mp.JoinableQueue()
time_proc = mp.Process(target=random_time, args=(time_queue,))
time_proc.daemon = True
time_proc.start()


random_time()
random_time()
random_time()
time_queue.join()
```
